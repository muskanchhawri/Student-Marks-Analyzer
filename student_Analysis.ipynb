# Student Marks Analyzer

## About the Project
Student Marks Analyzer is a Python project that analyzes student marks using NumPy and Pandas. It calculates total marks and percentages, assigns performance statuses, and provides subject-wise analysis.

## Features
- Store student marks in a NumPy array
- Calculate total marks and percentages
- Assign status: Excellent, Good, Pass, or Fail
- Find the highest and lowest total marks
- Calculate subject-wise average, highest, and lowest marks
- Create a student dataset using Pandas

## Technologies Used
- Python
- NumPy
- Pandas

## Dataset
The `students.csv` file contains the names of four students and their marks in four subjects.

## Project Files
- `student_marks_analyzer.py` — Main Python program
- `students.csv` — Student marks dataset
- `README.md` — Project documentation
